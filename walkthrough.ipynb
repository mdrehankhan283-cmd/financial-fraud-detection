```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
root = Path.cwd()
if not (root / 'train.py').exists(): root = root.parent
df = pd.read_csv(root / 'creditcard.csv')
df.shape, df.isna().sum().sum(), df.duplicated().sum()
```

```python
df.groupby('Class').agg(transactions=('Class','size'), median_amount=('Amount','median'))
```

```python
df.groupby('Class').Amount.plot.hist(bins=60, alpha=.5, logy=True, legend=True)
plt.title('Transaction amounts by class')
plt.show()
```
